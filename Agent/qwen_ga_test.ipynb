# 1. 测试原始Qwen节点速度

```python
import requests
import time

# 🔹 原生 DashScope 官方地址（无 GA）
url = "https://dashscope.aliyuncs.com/api/v1/services/aigc/text-generation/generation"

headers = {
    "Authorization": "sk-e05e5076e72e493998428e2d770e7a11",  # 替换为你的实际 API Key
    "Content-Type": "application/json"
}

payload = {
    "model": "qwen-turbo",
    "input": {
        "messages": [{"role": "user", "content": "test"}]
    },
    "parameters": {"max_tokens": 5}
}

session = requests.Session()  # 复用 TCP/TLS 连接
latencies = []

print("正在测试原生 DashScope API 延迟（直连，无 GA）...\n")

for i in range(5):
    start = time.perf_counter()
    try:
        response = session.post(url, json=payload, headers=headers, timeout=10)
        end = time.perf_counter()
        latency_ms = (end - start) * 1000
        latencies.append(latency_ms)
        status = response.status_code
        print(f"第 {i+1} 次: {latency_ms:.2f} ms (HTTP {status})")
        if status != 200:
            print(f"  ⚠️  错误响应: {response.text[:200]}...")
    except Exception as e:
        print(f"第 {i+1} 次失败: {e}")

if latencies:
    avg = sum(latencies) / len(latencies)
    p95 = sorted(latencies)[int(0.95 * len(latencies)) - 1] if len(latencies) > 1 else latencies[0]
    print(f"\n✅ 原生平均延迟: {avg:.2f} ms")
    print(f"✅ P95 延迟: {p95:.2f} ms")
else:
    print("❌ 所有请求失败，请检查网络或 API Key")
```
